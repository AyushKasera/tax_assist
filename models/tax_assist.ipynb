```python
## step up model example


examples = [
    {
        "input": "What is 80c and its rates",
        "output": "what is 80c and what are its rates",
    },
    {
        "input": "how to fill itr explain the types of its",
        "output": "how to fill itr and explain the types of itr",
    },
    {
        "input": "what is 80c,80d,80ccf and its rates",
        "output": "what is 80c and what are the rate of 80c and what is 80d and what are the rate of 80d and what is 80ccf and what are the rate of 80ccf",
    },
]
# We now transform these to example messages
example_prompt = ChatPromptTemplate.from_messages(
    [
        ("human", "{input}"),
        ("ai", "{output}"),
    ]
)
few_shot_prompt = FewShotChatMessagePromptTemplate(
    example_prompt=example_prompt,
    examples=examples,
)
prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            """You are an expert at world knowledge. Your task is to divide the question if need  and paraphrase each question to a more generic step-down question, which is easier to understand and classify parts of questions . Provide a single response Here are a few examples:""",
        ),
        # Few shot examples
        few_shot_prompt,
        # New question
        ("user", "{question}"),
    ]
)
```

```python
huggingface_repo_id="mistralai/Mistral-7B-Instruct-v0.3"
generate_queries_step_back = prompt | llm | StrOutputParser()
```

```python
question = "80d,tax"
response = generate_queries_step_back.invoke({"question": question})
response

```
